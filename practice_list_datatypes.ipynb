```python
# String: slicing, split(), join(), replace(), f-string formatting
#String Slicing
str="PHP is being attacted by cyber fraud"
print(str[:6])      #string slice from start to 5th index 6 is excluded
print(str[2:8])     #start from second index upto 7th index 8th index excluded
print(str[2:])      #start from 2nd index upto end
print(str[-1:])     #reverse index start from -1 upto end means 1 letter only
print(str[:-4])     #from start to -4 in reverse which is excluded upto before -4
print(str[-6:-2])   #from -6 to -3 because -2 is excluded
print(str[-2:-8])   #no output
print(str[8:2])     #no output
print(str[0:9])     #start to 8 because 9 is excluded
print(str[1:8:1])   #1 to 7 select each
print(str[1:8:2])   #step of 2 means move two index positions each time
print(str[1:8:3])   #step of 3 means move three index positions each time
print(str[1:8:-1])  #no output
print(str[::-1])    #reverse the string
print(str[2:9:-3])  #no output
print(str[2:-9])

#String Split
str1="sunflower is yellow"
words=str1.split(" ") #if we want to perform any operation on words we can use split
words.remove(words[1])
str2=""
for x in words:
  str2 += x+" "
print(str2.strip())
#one more way to join
str3 = " ".join(words)
print(str3)

#string replace
strr="sunflower is yellow"
strr.replace("sunflower","sun")

#String Formatting
quantity=2
name="sunflower"
print(f"i want {quantity} {name}")
print("i want {} {}".format(quantity,name))
```
